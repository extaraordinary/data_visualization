# **Exploring Data Visualization Techniques in Python**

*So, what is data actually?*

Strip away the buzzwords and data, at its core, is just this: numbers, timestamps, clicks, transactions, sensor readings — the exhaust of everything happening around us, captured and stored. The raw material of the digital age is in fact an ever-growing collection of facts, figures, and observations that capture the complexities of our world. 

In a world where information flows constantly around us, from social media feeds to economic indicators, this vast ocean of data holds the key to understanding the patterns and trends that shape our lives. However, raw data can be overwhelming and difficult to interpret. Data itself is not inherently meaningful. A spreadsheet with ten million rows doesn't know anything. he meaning shows up when you look at it right. Plot the same numbers as a scatter versus a heatmap and you'll notice completely different things; a cluster, a gap, or a trend that was invisible in the raw table. 

That's the whole job of data visualization, giving your eyes a shortcut past the noise so your brain can actually pattern-match by transforming complex datasets into clear, accessible visuals (such as charts, graphs, and maps) that reveal insights at a glance. Humans are wired to spot outliers in a chart in half a second; we're terrible at spotting them in a column of digits.

Below, we will delve into various visualization techniques, and provide examples using different Python libraries. This notebook is a working tour of creating effective data visualizations using three libraries with different philosophies, and when to reach for each one. We will explore different types of plots, customization techniques, and how to make your visualizations both informative and aesthetically pleasing.

• **Matplotlib:** A versatile, low-level library that can create nearly any type of plot with extensive control over every detail. Its powerful customization capabilities make it ideal for producing a wide range of complex visualizations, though this flexibility often comes with increased complexity, verbosity, and time investment.

• **Seaborn:** Built on Matplotlib, Seaborn is more user-friendly and excels in creating statistical plots with a modern, polished look. It simplifies common visualizations through concise code, attractive defaults, and stylish built-in themes, making it ideal for generating quick, professional-looking plots. Hence, Seaborn strikes a balance between ease of use and customization, offering just enough flexibility to tweak key elements without overwhelming options.

• **Plotly:** A powerful data visualization library, which is used for creating interactive, web-based graphs and charts in Python, R, and JavaScript. It is particularly favored for its ability to produce highly customizable and interactive plots that can be easily embedded in websites, dashboards, and reports.

# **1. Basic Plotting with Matplotlib**

## 1.1. Line Plot

Line plots are used to visualize data points connected by straight lines. They are particularly useful for displaying trends over time or relationships between continuous variables. Commonly employed in time series analysis, line plots help compare multiple datasets, identify patterns, or track changes over a period.

```python
plt.figure(figsize=(8, 6), dpi=300)     # create a figure (width = 8", height = 6"), increase dpi for higher resolution

x = [1, 2, 3, 4, 5]                     # initialize x data
y = [1, 5, 7, 12, 20]                   # initialize y data

plt.plot(x, y, label='Data')            # plot the data points with a line connecting them

plt.legend()                            # add a legend to identify the data series
plt.grid(True)                          # enable grid lines to enhance readability
plt.show()                              # display the plot
```

```python
plt.figure(figsize=(10, 6), dpi=300)

x = np.linspace(0, 10, 100)      # x is a sequence of 100 evenly spaced values from 0 to 10
y = np.sin(x)                    # y is the sine of each x value

plt.plot(x, y, label='Sine Wave', color='firebrick', linewidth=3)    # plot the sine wave

plt.title('Sin(x) Line Plot', fontsize=18, weight="bold")      # add title
plt.xlabel('X-axis', fontsize=16)                              # label the x and y axes
plt.ylabel('Y-axis', fontsize=16)
plt.xticks(fontsize=14)                                        # set the font size of the x and y axis tick labels
plt.yticks(fontsize=14)
plt.legend(fontsize=14)
plt.grid(False)
plt.show()
```

## 1.2. Scatter Plot

Scatter plots are a versatile tool used to examine the relationship between two variables by plotting individual data points on a Cartesian plane. Typically used in regression analysis, scatter plots are ideal for showing the spread, correlation, and potential causality between two variables. Scatter plots focus on the raw data and its distribution, making it easier to spot deviations and anomalies.

```python
plt.figure(figsize=(10, 6), dpi=300)

x = np.random.rand(50)           # generate 50 random points uniformly distributed between 0 and 1
y = np.random.rand(50)

plt.scatter(x, y, marker='h',    # marker shape
            s=150,               # marker size
            color='royalblue',   # marker color
            alpha=0.6)           # marker transparency (%)

plt.title('Scatter Plot', fontsize=16, weight='bold')
plt.xlabel('X-axis', fontsize=16)
plt.ylabel('Y-axis', fontsize=16)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.grid(True)
plt.show()
```

## 1.3. Bar Plot

Bar plots represent categorical data with rectangular bars. Each bar's length or height corresponds to the value or frequency of the category it represents, making it easy to compare different categories at a glance. Bar plots are particularly effective for displaying discrete data, such as counts or sums, across various categories, and can be oriented either vertically or horizontally.

```python
plt.figure(figsize=(10, 6), dpi=300)

categories = ['A', 'B', 'C', 'D']
values = [10, 15, 7, 20]
colors = ['khaki', 'dodgerblue', 'hotpink', 'darkseagreen']

plt.bar(categories, values, color=colors, width=0.5, edgecolor='black', linewidth=1.2)

plt.title('Bar Plot', fontsize=18, weight='bold')
plt.xlabel('Categories', fontsize=18)
plt.ylabel('Values', fontsize=18)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.grid(False)
plt.tight_layout()
plt.show()
```

As a side note, let’s use Seaborn to create a bar plot of the Penguins dataset. Seaborn makes it more visually appealing and user-friendly by offering features like easy hue adjustments and built-in aesthetic styles for a polished look. You can effortlessly highlight comparisons between categoreis and enhance clarity.

```python
plt.figure(figsize=(10, 6), dpi=300)

df = sns.load_dataset('penguins')      

sns.barplot(data=df, x='island', y='body_mass_g', hue='sex', palette='Set2')   # apply a color palette

plt.title('Bar Plot of Penguins Dataset', fontsize=18, weight='bold')
plt.xlabel('Island', fontsize=18)
plt.ylabel('Body Mass (g)', fontsize=18)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.grid(False)
plt.tight_layout()
plt.show()
```

## 1.4. Pie Chart

Pie charts are a popular and intuitive way to represent the proportionate contributions of different categories within a whole. Each slice of the pie corresponds to a category, with the size of the slice reflecting the proportion of that category relative to the total. Pie charts are particularly useful for illustrating the relative fractions of various components in a dataset, displaying simple comparisons among a few categories. They are often used in business and finance to display market shares, budget allocations, or survey results.

```python
plt.figure(figsize=(10, 8), dpi=300)

labels = ['Category A', 'Category B', 'Category C', 'Category D']  # pie slice labels
sizes = [15, 30, 45, 10]                                           # corresponding sizes for each slice

plt.pie(sizes, labels=labels, autopct='%1.1f%%',                   # show percentage of each slice
    startangle=140)                                                # rotate the start angle for better layout

plt.title('Pie Chart', fontsize=16, weight='bold')
plt.tight_layout()
plt.show()
```

Now, let’s have some fun with pizza pies—pun fully intended! Whether you are a fan of pineapple as a topping or prefer to leave it off, this code snippet allows you to slice and dice your data in a way that’s both insightful and engaging. 🍕

```python
plt.figure(figsize=(10, 10), dpi=300)

labels = ['Pepperoni', 'Mushrooms', 'Bell Peppers', 'Bacon', 'Olives', 'Onions', 'Extra Cheese']
sizes = [12, 15, 13, 25, 11, 8, 20]                                                   # sizes of each pie slice
colors = ['#FF6347', 'peru', 'forestgreen', 'firebrick', 'olive', 'lemonchiffon', 'gold']  # colors assigned to each slice
explode = (0, 0, 0, 0, 0, 0, 0.1)                                                     # highlight 'Extra Cheese'

def format_label(pct, all_values):                            # function to format labels with percentage and name
    total = sum(all_values)
    absolute = int(round(pct / 100. * total))
    label = labels[all_values.index(absolute)]
    return f'{label}\n{pct:.1f}%'

plt.pie(sizes, explode=explode, colors=colors, 
    autopct=lambda pct: format_label(pct, sizes),             # show percentage and label on each slice
    shadow=True,                                              # add shadow for a 3D effect
    startangle=140,                                           # rotate the start angle for a better layout
    wedgeprops={'edgecolor': 'black', 'linewidth': 1},        # black edge for slice separation
    textprops={'fontsize': 12, 'color': 'black'})             # style text labels

remaining_percentage = 100 - sum(sizes) / sum(sizes) * 100    # calculate the remaining percentage of pizza

plt.text(0, 0, f'Pizza Left: {remaining_percentage:.1f}%', ha='center', fontsize=14, weight='bold',
         bbox=dict(facecolor='white', edgecolor='black', boxstyle='round,pad=0.5'))

plt.title('Pizza Toppings Distribution', fontsize=16, weight="bold")
plt.show()
```

## 1.5. Hexbin Plot

Hexbin plots are used to visualize the density of points of a scatter plot in a 2D space. They are particularly useful for showing the relationship between two continuous variables when you have a large number of data points that overlap. Each bin represents the count of data points within that hexagon. The color intensity in each hexagon represents the number of data points it contains.

```python
plt.figure(figsize=(10, 8), dpi=300)

df = sns.load_dataset('tips')                        # load dataset

plt.hexbin(df['total_bill'], df['tip'], gridsize=30, cmap='Purples', mincnt=1)

cb = plt.colorbar(label='Counts')                    # add a colorbar with label
cb.ax.tick_params(labelsize=14)                      # set colorbar tick labels size
cb.ax.yaxis.label.set_size(16)                       # set colorbar label size

plt.title('Hexbin Plot of Total Bill vs. Tip', fontsize=16, weight='bold')
plt.xlabel('Total Bill ($)', fontsize=16)
plt.ylabel('Tip ($)', fontsize=16)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.show()
```

## 1.6. Radial Bar Chart

Radial bar charts visualize data in a circular format, where bars extend outward from a central point. They combine elements of traditional bar charts with a circular layout to represent data in a visually appealing way, which is useful for comparing categories that make up parts of a whole.

```python
fig, ax = plt.subplots(figsize=(10, 10), subplot_kw=dict(polar=True), dpi=300)

values = [7, 13, 22, 35, 44]
categories = ['A', 'B', 'C', 'D', 'E']

bars = ax.bar(np.linspace(0, 2 * np.pi, len(values), endpoint=False), values, width=0.4, color='slateblue', align='edge')

ax.set_xticks(np.linspace(0, 2 * np.pi, len(categories), endpoint=False))
ax.set_xticklabels(categories, fontsize=18, ha='center')

plt.title('Radial Bar Chart', fontsize=16, weight='bold')
plt.yticks(fontsize=16)
plt.show()
```

## 2.1. Histogram with KDE

Histograms show the distribution of a dataset, and Kernel Density Estimate (KDE) provides a smoothed curve to visualize the distribution. KDE is a way to estimate the probability density function of a continuous random variable. It is used for non-parametric analysis. 

```python
plt.figure(figsize=(10, 6), dpi=300)

df = sns.load_dataset('tips')      

sns.set_theme(style='whitegrid')                            # apply a Seaborn theme to all plots

sns.histplot(x=df['total_bill'], kde=True, color='peru')    # add a Kernel Density Estimate line
                                                            # switch x with y to rotate the histogram

mean_value = df['total_bill'].mean()                        # add a vertical line for the mean
plt.text(mean_value + 1, 37, f'Mean: {mean_value:.2f}', color='dimgrey', fontsize=14, weight='bold')
plt.axvline(mean_value, color='dimgrey', linestyle='--', linewidth=2)

plt.title('Histogram of Total Bill with KDE', fontsize=16, weight='bold')
plt.xlabel('Total Bill ($)', fontsize=16)
plt.ylabel('Count', fontsize=16)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.grid(False)
plt.show()
```

We will now apply this technique to the Tips dataset to visualize the distribution of total bills during lunch and dinner.

```python
plt.figure(figsize=(10, 6), dpi=300)

df = sns.load_dataset('tips')
palette = {"Lunch": "tomato", "Dinner": "midnightblue"}    # define a color palette

sns.set_style("whitegrid")                                 # set a Seaborn style

sns.histplot(data=df, x='total_bill', kde=True, hue='time', palette=palette, bins=15, edgecolor='black', linewidth=1.2)

plt.title('Distribution of Total Bill by Time', fontsize=16, weight='bold')
plt.xlabel('Total Bill ($)', fontsize=16)
plt.ylabel('Count', fontsize=16)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.grid(False)
plt.show()
```

## 2.2. Ridge Plot (Joy Plot)

Ridge plots, also known as joy plots, display the distribution of a continuous variable across different categories or conditions. They are particularly useful for visualizing the overlap and distribution of multiple distributions in a compact and comparative manner. Ridge plots stack density plots or histograms of different groups on top of each other, often with transparent or partially transparent fill colors, to show how the distributions vary across categories.

```python
plt.figure(figsize=(10, 6), dpi=300)

df = sns.load_dataset('tips')
palette = {"Lunch": "indianred", "Dinner": "mediumaquamarine"}

sns.kdeplot(data=df, x='tip', hue='time', fill=True, common_norm=False, alpha=0.6, palette=palette)
plt.title('Ridge Plot of Tips by Time of Day', fontsize=18, weight='bold')
plt.xlabel('Tip Amount ($)', fontsize=18)
plt.ylabel('Density', fontsize=18)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.grid(False)
plt.show()
```

## 2.3. Box Plot

Box plots, also known as whisker plots, are a standardized way of displaying the distribution of data based on a five-number summary: minimum, first quartile (Q1), median, third quartile (Q3), and maximum. The "box" in the plot represents the interquartile range (IQR), which contains the middle 50% of the data, with a line inside the box indicating the median. The "whiskers" extend from the box to the smallest and largest values within 1.5 times the IQR, and any data points outside this range are considered outliers and are often plotted as individual dots. Box plots are particularly useful for comparing distributions across multiple groups and identifying potential outliers. They provide a clear visualization of the spread, central tendency, and variability of data, making them a valuable tool in exploratory data analysis.

```python
plt.figure(figsize=(10, 6), dpi=300)

df = sns.load_dataset('tips')

sns.boxplot(x='day', y='total_bill', data=df, hue='sex', palette='Set1')   # apply a pre-defined color palette

plt.title('Box Plot of Total Bill by Day', fontsize=16, weight='bold')
plt.xlabel('Day of the Week', fontsize=16)
plt.ylabel('Total Bill ($)', fontsize=16)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.show()
```

## 2.4. Violin Plot

Violin plots combine features of box plots and density plots to show the distribution and the probability density of a continuous variable across different categories. They display the data's distribution, central tendency, and variability by using mirrored density plots, which reveal the shape and spread of the data. This makes them ideal for comparing distributions across categories and understanding the density of data points. Violin plots are especially useful in exploratory data analysis.

```python
plt.figure(figsize=(10, 8), dpi=300)

df = sns.load_dataset('penguins')

sns.violinplot(x='species', y='flipper_length_mm', data=df, palette='muted')

plt.title('Violin Plot of Flipper Length by Penguin Species', fontsize=16, weight="bold")
plt.xlabel('Species', fontsize=16)
plt.ylabel('Flipper Length (mm)', fontsize=16)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.show()
```

## 2.5. Swarm Plot

Swarm plots are a type of scatter plot that displays individual data points with a jittered arrangement to avoid overlapping. This technique is ideal for displaying all observations and understanding the distribution and density of categorical data.

```python
plt.figure(figsize=(10, 6), dpi=300)

df = sns.load_dataset("penguins")

sns.swarmplot(x='species', y='flipper_length_mm', data=df, s=5, hue='sex', palette='Accent')

plt.title('Swarm Plot of Flipper Length by Penguin Species', fontsize=16, weight="bold")
plt.xlabel('Species', fontsize=16)
plt.ylabel('Flipper Length (mm)', fontsize=16)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.legend(fontsize=14)
plt.show()
```

## 2.6. Heatmap

Heat maps are used to represent the intensity of values across a matrix. By using color gradients, heat maps provide a clear understanding of data density or magnitude, making them ideal for visualizing correlation matrices, time series, and spatial distributions. Adding annotations can help identify specific values directly on the plot.

```python
plt.figure(figsize=(8, 8), dpi=300)

df = sns.load_dataset('penguins')
penguins = df.dropna()                            # drop rows with missing values

corr_matrix = penguins.corr(numeric_only=True)    # calculate the correlation matrix

heatmap = sns.heatmap(corr_matrix, 
    annot=True, annot_kws={'size': 10},           # annotate each cell with correlation coefficient
    linewidths=1, linecolor='white',              # width and color of the lines separating cells
    cmap='cividis', cbar_kws={'shrink': 0.75},    # color map for positive and negative correlations
    square=True)                                  # make each cell square-shaped for a uniform look                 

cbar = heatmap.collections[0].colorbar            # get the color bar object from the heatmap
cbar.ax.tick_params(labelsize=10)

plt.xticks(fontsize=10)                           
plt.yticks(fontsize=10)
plt.title('Heatmap of Penguins Dataset', fontsize=12, fontweight='bold', pad=20)
plt.show()
```

## 2.7. Pair Plot

Pair plots explore relationships between multiple variables by plotting pairwise relationships in a dataset. They create a matrix of scatter plots, where each plot represents the relationship between a pair of variables, allowing for a comprehensive view of how different features interact with one another. Along the diagonal, histograms or KDE plots illustrate the individual distribution of each variable. By visualizing these relationships in a grid format, pair plots enable a quick and intuitive understanding of complex data structures, making them an invaluable tool for performing tasks such as feature selection and multivariate analysis.

```python
plt.rcParams['figure.dpi'] = 300

df = sns.load_dataset('penguins')

sns.pairplot(df, hue='species',                            # differentiate by species
    palette='Set1',                                        # use a pleasant color palette
    diag_kind='kde',                                       # KDE plots on the diagonal for smooth density estimation
    height=3,                                              # size of the subplots
    markers=['o', 's', 'D'], plot_kws={'alpha':0.8})       # marker shape and transparency for better visibility

plt.suptitle('Pair Plot of Penguins Dataset', fontsize=22, fontweight='bold', y=1.02)
plt.show()
```

```python
plt.rcParams['figure.dpi'] = 300

df = sns.load_dataset('penguins')

sns.pairplot(df, hue='species', palette='Set1', kind='kde', height=3, markers=['o', 's', 'D'], plot_kws={'alpha':0.8})

plt.suptitle('Pair Plot of Penguins Dataset', fontsize=22, fontweight='bold', y=1.02)
plt.show()
```

## 2.8. FacetGrid Plot

FacetGrid visualizes complex datasets by creating a grid of subplots based on multiple categorical variables. It maps one or more dimensions of data to rows, columns, and even hue, making it easy to explore relationships within subsets of data. Each subplot, or facet, represents a specific subset defined by unique combinations of the categorical variables, enabling the comparison of patterns and trends across these subsets. FacetGrid is particularly useful for multi-dimensional data, where understanding the interactions between different variables is crucial.

```python
plt.rcParams['figure.dpi'] = 300

df = sns.load_dataset("tips")
palette = {"Male": "seagreen", "Female": "indigo"}  # define a color palette for male and female

g = sns.FacetGrid(df, col="time", row="sex", 
                  margin_titles=True, height=4.5, aspect=1.4, despine=False, palette=palette)

g.map_dataframe(sns.histplot, x="total_bill", kde=True, hue="sex", palette=palette)  # map histogram plot to the grid

g.set_axis_labels("Total Bill ($)", "Count", fontsize=20)
g.set_titles(col_template="{col_name} Time", row_template="{row_name} Gender", size=20)
g.set_xticklabels(fontsize=20)
g.set_yticklabels(fontsize=20)
plt.suptitle('FacetGrid Distribution of Total Bill by Time and Gender', fontsize=22, weight='bold', y=1.05)
plt.subplots_adjust(top=0.9)
plt.show()
```

## 2.9. Joint Plot with Multiple Visualizations

Joint plots combine two key features into a single plot: a scatter plot and corresponding marginal histograms or density plots, thereby providing a comprehensive view of the relationship between two variables by displaying their joint distribution, along with their individual distributions. The scatter plot component highlights correlations or patterns between the variables, while the marginal plots on the axes provide insights into the distribution of each variable independently. Joint plots can be enhanced with regression lines, KDE, or even hexbin plots for larger datasets.

```python
plt.rcParams['figure.dpi'] = 300

df = sns.load_dataset('tips')
sns.set_style("whitegrid")    # set the style of the plot

g = sns.jointplot(x='total_bill', y='tip', data=df, kind='reg', height=8, color='slateblue',
    scatter_kws={'s': 100, 'alpha': 0.6, 'color': 'olive', 'edgecolor': 'black', 'linewidths': 1})

g.plot_joint(sns.regplot, 
    scatter_kws={'s': 100, 'alpha': 0.6, 'color': 'olive', 'edgecolor': 'black', 'linewidths': 1},
    line_kws={'color': 'red', 'lw': 2.5, 'linestyle': '--'})

g.fig.suptitle('Joint Plot with Regression Line', fontsize=16, y=1.02, weight='bold')
g.set_axis_labels('Total Bill ($)', 'Tip ($)', fontsize=16)
g.ax_joint.tick_params(axis='both', which='major', labelsize=14)
plt.show()
```

## 3.4. Interactive Choropleth Map

Choropleth maps represent geographic data through color-coded regions, effectively illustrating variations in data across different locations. Using Plotly's px.choropleth function, users can map data values to geographical regions, such as countries, states, or any predefined geospatial regions. The color intensity or gradient represents data values, such as population density, GDP, or election results, providing an intuitive and visually appealing way to convey complex information. Plotly’s choropleth maps are highly customizable, supporting various color scales, map projections, and interactive features like hover tooltips and animations. These dynamic, interactive maps are an invaluable tool for data exploration, storytelling and presentation in diverse fields including economics, politics, and environmental science.

```python
data = px.data.gapminder().query("year == 2007")

fig = px.choropleth(                                        # create the choropleth map
    data,
    locations="iso_alpha",
    color="gdpPercap",
    hover_name="country",
    color_continuous_scale=px.colors.sequential.Plasma,
    title='World GDP Per Capita in 2007',
    projection="natural earth",                             # adjust projection for better visual appeal
    labels={'gdpPercap': 'GDP Per Capita'}                  # label for the color scale
)

fig.update_layout(                                          # update layout for improved aesthetics
    title={
        'text': "World GDP Per Capita in 2007",
        'x': 0.5,
        'xanchor': 'center',
        'font': {'size': 20, 'family': 'Arial'}
    },
    geo=dict(
        showcoastlines=True,
        coastlinecolor='Black',
        showland=True,
        landcolor='lightgray',
        showocean=True,
        oceancolor='lightblue',
        showframe=False
    ),
    coloraxis_colorbar=dict(
        title='GDP Per Capita',
        tickvals=[1000, 10000, 50000],  
        ticktext=['$1,000', '$10,000', '$50,000']           
    ),
    margin=dict(t=50, b=0, l=0, r=0)                       # adjust margins for better layout
)

fig.show()
```

## 3.5. Treemap

Tree maps are a type of hierarchical data visualization that display information as nested rectangles, where each rectangle's size represents a quantitative measure, such as sales or population, and colors can indicate additional attributes. This compact visualization method efficiently shows proportions and relationships within large datasets, making it easy to understand the relative magnitude of different categories. Tree maps are particularly useful for financial analysis, organizational structures, and market research, providing a clear, interactive way to explore and interpret complex data hierarchies.

```python
data = {
    'Category': ['Electronics', 'Electronics', 'Furniture', 'Furniture', 'Office Supplies', 'Office Supplies'],
    'Sub-Category': ['Phones', 'Laptops', 'Chairs', 'Tables', 'Pens', 'Binders'],
    'Sales': [15000, 20000, 12000, 18000, 8000, 6000]
}

df = pd.DataFrame(data)

fig = px.treemap(                                             # create the tree map
    df,
    path=['Category', 'Sub-Category'],                        # define hierarchy
    values='Sales',                                           # size of each rectangle based on sales
    color='Sales',                                            # color of each rectangle based on sales
    color_continuous_scale=px.colors.sequential.Cividis,      # enhance color scale
    title='Sales Distribution by Category and Sub-Category',  # Title
    labels={'Sales': 'Sales ($)', 'Category': 'Product Category', 'Sub-Category': 'Sub-Category'}
)

fig.update_layout(
    title={
        'text': 'Sales Distribution by Category and Sub-Category',
        'x': 0.5,
        'xanchor': 'center',
        'font': {'size': 20, 'family': 'Arial'}
    },
    margin=dict(t=50, b=0, l=0, r=0),                         # adjust margins for a cleaner look
    paper_bgcolor='rgba(255, 255, 255, 0.8)',                 # background color
    font=dict(family='Arial', size=12)                        # font settings
)

fig.show()
```

## 4.1. Subplots in Matplotlib

You can combine multiple plots in a single figure using subplots.

```python
fig, axs = plt.subplots(2, 2, figsize=(14, 10), tight_layout=True)  # Subplots

tips = sns.load_dataset('tips')

# Prepare data for pie chart
tips_summary = tips.groupby('day')['total_bill'].sum()
sizes = tips_summary
labels = tips_summary.index
colors = sns.color_palette('Set2')[0:len(labels)]

# Line Plot: Total Bill by Day
sns.lineplot(data=tips, x='day', y='total_bill', estimator='mean', marker='o', linestyle='-', color='tab:orange', 
    linewidth=3, markersize=15, ax=axs[0, 0])
axs[0, 0].set_title('Average Total Bill by Day', fontsize=14, fontweight='bold')
axs[0, 0].set_xlabel('Day', fontsize=12)
axs[0, 0].set_ylabel('Average Total Bill ($)', fontsize=12)
day_order = tips['day'].unique()
axs[0, 0].set_xticks(range(len(day_order)))
axs[0, 0].set_xticklabels(day_order, fontsize=10)
axs[0, 0].grid(True, linestyle='--', alpha=0.5)

# Scatter Plot: Total Bill vs Tip
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', palette='Set2', edgecolor='k', alpha=0.8, s=100, ax=axs[0, 1])
sns.regplot(x='total_bill', y='tip', data=tips, scatter=False, ax=axs[0, 1], color='black', line_kws={'linewidth': 2})
axs[0, 1].set_title('Total Bill vs Tip with Sex Hue', fontsize=14)
axs[0, 1].set_xlabel('Total Bill ($)', fontsize=12)
axs[0, 1].set_ylabel('Tip ($)', fontsize=12)
axs[0, 1].legend(title='sex')
axs[0, 1].grid(linestyle='--', alpha=0.5)

# Bar Plot: Average Tip by Day
sns.barplot(data=tips, x='day', y='tip', hue='sex', ax=axs[1, 0], palette='Set2', edgecolor='black')
axs[1, 0].set_title('Average Tip by Day', fontsize=14)
axs[1, 0].set_xlabel('Day', fontsize=12)
axs[1, 0].set_ylabel('Average Tip ($)', fontsize=12)
axs[1, 0].grid(axis='y', linestyle='--', alpha=0.5)

# Pie Chart: Total Bill Distribution by Day
def autopct_with_fontsize(pct):
    return f'{pct:.1f}%'
wedges, texts, autotexts = axs[1, 1].pie(sizes, labels=labels, colors=colors, 
                                          autopct=autopct_with_fontsize, startangle=140,  wedgeprops=dict(width=0.7))
for text in texts:
    text.set_fontsize(14)
for autotext in autotexts:
    autotext.set_fontsize(14)
axs[1, 1].set_title('Total Bill Distribution by Day', fontsize=14)

plt.suptitle('Combining Multiple Visualizations Using Tips Dataset', fontsize=20, fontweight='bold')
plt.show()
```

## 4.2. Dashboards with Plotly

Plotly allows for the creation of interactive dashboards by combining multiple plots.

```python
import plotly.graph_objs as go
from plotly.subplots import make_subplots

penguins = sns.load_dataset('penguins').dropna()        # drop rows with missing values

fig = make_subplots(                                    # create subplots
    rows=2, cols=2,
    specs=[[{'type': 'xy'}, {'type': 'xy'}],
           [{'type': 'xy'}, {'type': 'domain'}]],       # 'domain' is for pie charts
    subplot_titles=('Scatter Plot', 'Bar Plot', 'Line Plot', 'Pie Chart'),
    vertical_spacing=0.2,                               # adjust spacing between plots
    horizontal_spacing=0.15)                            # adjust spacing between plots

# Scatter Plot
fig.add_trace(
    go.Scatter(
        x=penguins['flipper_length_mm'], y=penguins['body_mass_g'], mode='markers',
        marker=dict(color=penguins['species'].astype('category').cat.codes, colorscale='Viridis', size=10),
        name='Body Mass vs Flipper Length', text=penguins['species'], showlegend=False),
    row=1, col=1)

# Bar Plot
avg_flipper_length = penguins.groupby('species')['flipper_length_mm'].mean().reset_index()
fig.add_trace(
    go.Bar(
        x=avg_flipper_length['species'], y=avg_flipper_length['flipper_length_mm'],
        name='Average Flipper Length', marker=dict(color=['indigo', 'teal', 'gold'])),
    row=1, col=2)

# Line Plot
avg_body_mass_per_island = penguins.groupby('island')['body_mass_g'].mean().reset_index()
fig.add_trace(
    go.Scatter(
        x=avg_body_mass_per_island['island'], y=avg_body_mass_per_island['body_mass_g'], mode='lines+markers', 
        name='Average Body Mass by Island', line=dict(color='hotpink', width=3), marker=dict(size=15)),
    row=2, col=1)

# Pie Chart
species_counts = penguins['species'].value_counts()
fig.add_trace(
    go.Pie(
        labels=species_counts.index, values=species_counts.values, name='Species Distribution', hole=0.3,
        textinfo='label+percent', marker=dict(colors=['indigo', 'gold', 'teal']), title=dict(text='Species', font_size=16)),
    row=2, col=2)

fig.update_layout(
    height=900, width=1200, title_text="Penguins Dataset: Subplots with Different Plot Types", title_x=0.5,
    font=dict(size=14),                                           # set font size for all text in the layout
    xaxis_title='Flipper Length (mm)',                            # x-axis title for scatter plot
    yaxis_title='Body Mass (g)',                                  # y-axis title for scatter plot
    xaxis2_title='Species',                                       # x-axis title for bar plot
    yaxis2_title='Average Flipper Length (mm)',                   # y-axis title for bar plot
    xaxis3_title='Island',                                        # x-axis title for line plot
    yaxis3_title='Average Body Mass (g)',                         # y-axis title for line plot
    showlegend=False)

fig.show()
```

## **Data Query**

A data query is a powerful tool used to extract specific subsets of data from a larger dataset based on defined criteria. By applying a query, users can filter, sort, and analyze data to focus on particular aspects or subsets that are of interest. For instance, in a dataset containing global socio-economic indicators, a query can be used to isolate data for a specific country, such as the United Kingdom, or to select records from a particular year or range of years. This targeted approach not only simplifies data analysis by reducing the volume of data to only what's relevant but also enhances the relevance of insights drawn from the data. Queries are essential in data analysis for generating meaningful insights, identifying trends, and making data-driven decisions.

We use the Gapminder dataset bundled with Plotly (`plotly.express.data.gapminder()`), which provides annual data on global development metrics. This dataset includes various columns that track socio-economic indicators such as life expectancy, GDP per capita, and population figures for different countries across multiple years.

---

### **Key Features of the Dataset:**

• country: The name of the country.

• year: The year of the recorded data.

• continent: The continent on which the country is located.

• lifeExp: Life expectancy at birth, representing the average number of years a newborn is expected to live.

• gdpPercap: GDP per capita, indicating the economic output per person.

• pop: Population size of the country.

### **Purpose of the Analysis:**

The goal is to analyze trends in the United Kingdom’s socio-economic indicators over time. By examining metrics such as life expectancy and GDP per capita, we can gain insights into how the UK's development has progressed and identify patterns or anomalies in the data.

With this dataset, we can generate various visualizations to explore:

• Trends in life expectancy over the years.

• Changes in GDP per capita and its relationship with life expectancy.

• Comparative analysis of different countries if needed.

---

We first filter the dataset to focus specifically on the United Kingdom by using the query method: gapminder.query('country == "X"'). 

```python
# Use Plotly's bundled Gapminder dataset (no network request needed, and
# avoids relying on the now-dead bit.ly redirect used in the original version).
gapminder = px.data.gapminder()

df_uk = gapminder.query('country == "United Kingdom"')      # filter data

plt.figure(figsize=(10, 6), dpi=300)

plt.plot(df_uk['lifeExp'], df_uk['gdpPercap'], marker='o', linestyle='-', lw=3, color='darkblue', 
         markersize=12, markerfacecolor='darkblue', markeredgecolor='black', markeredgewidth=1)

plt.title('GDP Per Capita vs Life Expectancy in the United Kingdom', fontsize=16, fontweight='bold')
plt.xlabel('Life Expectancy (Years)', fontsize=16)
plt.ylabel('GDP Per Capita (USD)', fontsize=16)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

An exploration of Seaborn's Flights dataset to analyze and visualize trends in airline passenger traffic is covered below. This dataset provides a rich view of passenger counts over time, spanning from 1949 to 1960, and is ideal for examining both long-term trends and seasonal patterns.

```python
plt.figure(figsize=(10, 8), dpi=300)

sns.set(style='whitegrid')

flights = sns.load_dataset('flights')
may_flights = flights.query("month == 'Jan'")

sns.lineplot(data=may_flights, x='year', y='passengers', color='dodgerblue', linewidth=4, marker='o', markersize=13)

plt.title('Number of Passengers in January (1949-1960)', fontsize=18, fontweight='bold')
plt.xlabel('Year', fontsize=18)
plt.ylabel('Number of Passengers', fontsize=18)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,8), dpi=300)

flights_wide = flights.pivot(index='year', columns='month', values='passengers')

sns.lineplot(data=flights_wide, linewidth=3, palette='husl', dashes=False, markers=False, markerfacecolor='black')

plt.title('Number of Passengers per Year', fontsize=18, fontweight='bold')
plt.xlabel('Year', fontsize=18)
plt.ylabel('Number of Passengers', fontsize=18)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(title='Month', title_fontsize='16', fontsize='14', loc='upper left', bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```

**Handling a dataset passed in "long-form" mode:** When the lineplot function encounters multiple data points for the same x-value (in this case, the same year), it aggregates these points. By default, Seaborn calculates the mean of the y-values (e.g., the number of passengers) for each unique x-value (year).

**95% Confidence Interval (CI):** Along with the mean, Seaborn also calculates a confidence interval, typically a 95% CI, which provides a range within which the true mean is expected to fall, with 95% certainty. This interval is visualized as a shaded area around the line, representing the uncertainty or variability in the data.

```python
plt.figure(figsize=(12, 8), dpi=300)

sns.lineplot(data=flights, x='year', y='passengers', linewidth=3, marker='o', markersize=8, color='blue')

plt.title('Annual Number of Airline Passengers (1949-1960)', fontsize=20, fontweight='bold')
plt.xlabel('Year', fontsize=18)
plt.ylabel('Number of Passengers', fontsize=18)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
